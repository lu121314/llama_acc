# 网页 UI
先在本工程的 Jupyter **终端**运行 `bash run.sh --mode ui`。服务启动后会打印 HTTP 访问地址，点击链接打开；部分 Jupyter 版本需要 `Ctrl+单击`。保持终端打开，按 `Ctrl+C` 停止服务。
下面单元可将同一服务嵌入 Notebook，不创建第二个硬件会话。默认端口为 8088；如果启动时修改了端口，请同步修改下面代码。
若 Jupyter 由 HTTPS 打开导致浏览器阻止混合内容，请直接点击终端打印的 HTTP 地址，在新标签页访问。

```python
from IPython.display import HTML, display
display(HTML("""<p><a id="llama-ui-link" target="_blank">在新标签页打开网页 UI</a></p>
<iframe id="llama-ui-frame" title="Llama FPGA 网页界面" style="width:100%;height:900px;border:0"></iframe>
<script>
const url = "http://" + window.location.hostname + ":8088/";
document.getElementById("llama-ui-link").href = url;
document.getElementById("llama-ui-frame").src = url;
</script>"""))
```
